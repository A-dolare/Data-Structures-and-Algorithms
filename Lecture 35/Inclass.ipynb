```python
from collections import deque
class Solution:
    def bfs(self, adj):
        # code here
        v = len(adj)
        q = deque()
        visited = [False] * v
        ans = []
        
        q.append(0)
        
        while len(q) > 0:
            # remove
            curr = q.popleft()
            
            # print
            visited[curr] = True
            ans.append(curr)
            
            # add adjacents
            neighbours = adj[curr]
            for vert in neighbours:
                if not visited[vert]:
                    visited[vert] = True
                    q.append(vert)
                    # ans.append(vert)
                # print(q, ans, visited)
                    
        return ans
```

```python
class Solution:
    def checkPath(self, V, edges, src, dest):
        # Code here
        # if V < 3:
        #     return True
        
        visited = [False] * V
        
        # make an adjacency list
        adj = [[] for _ in range(V)]
        
        for edge in edges:
            adj[edge[0]].append(edge[1])
            adj[edge[1]].append(edge[0])
            
            
        def dfs(curr, dest, visited, adj):
            # remove
            visited[curr] = True
            
            # print
            if curr == dest:
                return True
            
            # add adjacents
            ngbrs = adj[curr]
            for vert in ngbrs:
                if not visited[vert]:
                    if dfs(vert, dest, visited, adj):
                        return True
                    
                    
            return False
            
        if dfs(src, dest, visited, adj):
            return True
        return False
                        
        
```
