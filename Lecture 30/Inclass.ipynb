```python
'''
Definition for Node
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None
'''

class Solution:
    def lca(self, root, n1, n2):
        # code here
        
        if not root:
            return None
            
        if root.data == n1 or root.data == n2:
            return root
            
        lst = self.lca(root.left, n1, n2)
        rst = self.lca(root.right, n1, n2)
        
        if lst is None and rst is None:
            return None
        if lst and rst:
            return root
        if lst is not None:
            return lst
        if rst is not None:
            return rst
        
        
        
```

```python
'''
Definition for Node
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None
'''

class Solution:
    def insert(self, root, key):
        # code here 
        cur = root
        insert = Node(key)
        while True:
            
            if cur.data > key:
                if not cur.left:
                    cur.left = insert
                    break
                else:
                    cur = cur.left 
                
            elif cur.data < key:
                if not cur.right:
                    cur.right = insert
                    break
                else:
                    cur = cur.right
            
        return root
        
```

```python
'''
Definition for Node
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None
'''

class Solution:
    def search(self, root, key):
        # code here
        
        # Base Cond
        if not root:
            return False
            
        if root.data > key:
            return self.search(root.left, key)
            
        elif root.data < key:
            return self.search(root.right, key)
            
        else:
            return True
            
        
```
