Merge Sort



```python
class Solution:
 
    def mergeSort(self, arr, l, r):
        #code here
        
        if l >= r:
            return
        
        # n = len(arr)
        mid = (l + r) // 2
        
        # print(l, end=' ')
        # print(r, end=' ')
        # print(mid)

        self.mergeSort(arr, l, mid)
        self.mergeSort(arr, mid + 1, r)
                
        self.merge(arr, l, r, mid)
        
        # return arr
 
        
    def merge(self, arr, l, r, mid):
        
        # split between arr1[l, ...., mid] and arr2[mid + 1, ...., r]
        
        L = [0] * (mid - l + 1)
        R = [0] * (r - mid)
        
        n1 = mid - l + 1
        n2 = r - mid
        
        for i in range(n1):
            L[i] = arr[l + i]
        for i in range(n2):
            R[i] = arr[mid + 1 + i]
            
        # print(L, R, sep= ' ' )
        
        # print(R)
        
        # L[i] < R[j] --> add L[i] in arr[] at the index k
        # else (L[i] >= R[j]) --> add R[j] in arr[] at the index k
        
        k = l
        i = 0
        j = 0 
        
        while i < n1 and j < n2:
            if L[i] < R[j]:
                arr[k] = L[i]
                i += 1
                k += 1
            else:
                arr[k] = R[j]
                j += 1
                k += 1
                
        while i < n1:
            arr[k] = L[i]
            i += 1
            k += 1
        while j < n2:
            arr[k] = R[j]
            j += 1
            k += 1
```

Counting Sort

```python

n = int(input())
arr = list(map(int, input().split()))
n = len(arr)
# print(n, arr)

maxi = max(arr)

# print(max)

freq = [0] * (maxi + 1)
for i in range(n):
    freq[arr[i]] += 1
    
# print(freq)

# Prefix Sum of Freq array
pref = [0] * len(freq)
for i in range(len(freq)):
    pref[i] = pref[i - 1] + freq[i]
# print(pref)

ans = [0] * n

for i in range(n):
    ele = arr[i]
    idx = pref[ele] - 1
    ans[idx] = ele
    pref[ele] -= 1

print(ans)
```
