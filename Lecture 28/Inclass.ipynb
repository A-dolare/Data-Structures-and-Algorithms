         1
        / \
       2   3
      / \
     4   5   

| Step | Node Visited | `lsh` (Left Subtree Height) | `rsh` (Right Subtree Height) | `self.ans` Update (max diameter) | Return Value (Height) |
|:----:|:------------:|:---------------------------:|:----------------------------:|:--------------------------------:|:---------------------:|
| 1 | `4` (leaf) | 0 (left is None) | 0 (right is None) | `max(0, 0+0) = 0` | `1 + max(0,0) = 1` |
| 2 | `5` (leaf) | 0 (left is None) | 0 (right is None) | `max(0, 0+0) = 0` | `1 + max(0,0) = 1` |
| 3 | `2` | 1 (from node 4) | 1 (from node 5) | `max(0, 1+1) = 2` | `1 + max(1,1) = 2` |
| 4 | `3` (leaf) | 0 (left is None) | 0 (right is None) | `max(2, 0+0) = 2` | `1 + max(0,0) = 1` |
| 5 | `1` (root) | 2 (from node 2) | 1 (from node 3) | `max(2, 2+1) = 3` | `1 + max(2,1) = 3` |

```python
# Postorder traversal
'''
Definition for Node
class Node:
    def __init__(self, val):
        self.data = val
        self.right = None
        self.left = None
'''

class Solution:
    def __init__(self):
        self.ans = 0  # Instance variable to store max diameter

    def diameter(self, root):
        self.ans = 0  # Reset for each test case
        self.height(root)
        return self.ans

    def height(self, root):
        if root is None:
            return 0  # Height of empty tree is 0

        # Recursively get heights of left and right subtrees
        lsh = self.height(root.left)
        rsh = self.height(root.right)

        # Update diameter: path through this node = left_height + right_height (in edges)
        self.ans = max(self.ans, lsh + rsh)

        # Return height of current subtree (in nodes)
        return 1 + max(lsh, rsh)   
```

```python
''' 
class Node:

    def __init__(self, val):
        self.data = val
        self.right = None
        self.left = None 
'''
from collections import deque
class Solution:
    def __init__(self):
        self.ans = 0

    def leftView(self, root):
        # code here
        
        if root is None:
            return []
        
        q = deque()

        q.append(root)

        level = -1

        size = 1
        
        ans = []

        while len(q) != 0:
            # print(size)
            
            curr_lev = []
            while size > 0:

                node = q.popleft()
                
                curr_lev.append(node.data)

                if node.left != None:
                    q.append(node.left)
                if node.right != None:
                    q.append(node.right)

                size -= 1
            level += 1

            size = len(q)
            
            ans.append(curr_lev[0])

        return ans
        
        
```

```python
'''
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None
'''
from collections import deque
class Solution:
    def topView(self, root):
        # code here
        
        if root is None:
            return []
        
        q = deque()
        q.append((root, 0))
        
        h_map = {}
        h_map[0] = root.data
        
        size = 1
        level = -1
        
        min_hd = 0
        max_hd = 0
        
        ans = []
        # print(min_hd, max_hd)
        # return []
        
        # R P Lc Rc
        while q:
            # print(q)
            while size > 0:
                
                node, hd = q.popleft()
                
                if node.left != None:
                    q.append((node.left, hd - 1))
                    
                if node.right != None:
                    q.append((node.right, hd + 1))
                    
                if hd < min_hd:
                    min_hd = hd
                
                elif hd > max_hd:
                    max_hd = hd
                    
                if hd not in h_map:
                    h_map[hd] = node.data
                
                # print(size)
                size -= 1
                
            size = len(q)
        # print(h_map)
        
        for i in range(min_hd, max_hd + 1):
            ans.append(h_map[i])
            
        return ans
```
