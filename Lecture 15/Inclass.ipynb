```python
class Solution:
    def singleNumber(self, nums: List[int]) -> int:

        n = len(nums)
        ans = 0
        
        for i in range(32):
            count = 0
            
            for j in range(n):
                
                if nums[j] >> i & 1 != 0:
                    count += 1
                
                    
            if count % 3 != 0:
                ans = ans | (1 << i)

        if ans >= 2 ** 31:
            ans -= 2 ** 32
                    
        return ans
```
