```python
from collections import deque
class Solution:
    def removeDuplicates(self, s: str) -> str:

        # print(len(s))

        n = len(s)
        q = deque()
        q.append(s[0])
        for i in range(1, n):
            if q and s[i] == q[-1]:
                q.pop()
            else:
                q.append(s[i])
        #     print(q)
        # print(q)
        # return ''
        return ''.join(q)
        
```

```python
from collections import deque
class Solution:
    def removeStars(self, s: str) -> str:
        n = len(s)
        q = deque()
        q.append(s[0])
        i = 1
        while i < n:
            if q and s[i] == '*':
                q.pop()
                i += 1
            else:
                q.append(s[i])
                i += 1
            # print(q, s[i + 1])

            # if i + 1 < n and s[i + 1] == '*':
            #     i += 2
            #     continue

        return ''.join(q)
```

```python

class Solution:
    def removeDuplicates(self, s: str, k: int) -> str:
        # n = len(s)
        q = []
        
        for char in s:
            if q and q[-1][0] == char and q[-1][1] == k - 1:
                q.pop()
            elif q and q[-1][0] == char:
                q[-1][1] += 1
            else:
                q.append([char, 1])
            # print(q)

        # char, freq = q.pop()
        # print(char, freq)
        s = ''

        while q:
            char, freq = q.pop()
            for i in range(freq):
                s += char
        
        return s[::-1]

        
```
