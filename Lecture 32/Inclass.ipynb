```python
# Using Max_heap 
class Solution:
    def heapSort(self, arr):
        #code here
        n = len(arr)
        self.temp = arr[:]
        
        arr.clear()
        
        self.build_heap(n)
        
        while len(self.temp) > 0:
            
            arr.append(self.temp[0])
            
            # self.temp[0] = self.temp[len(self.temp) - 1]
            
            item = self.temp.pop()
            # print(self.temp)
            
            if len(self.temp) > 0:
                self.temp[0] = item
                
                self.max_heapify(len(self.temp), 0)
            
        arr.reverse()
        return arr
        
    def build_heap(self, n): # Build_Heap is done on a unsorted array to bring the max/min element to the root
        
        for i in range((n // 2) - 1, -1, -1):
            
            self.max_heapify(n, i)
        
    def max_heapify(self, n, i): # Heapify is done on a index
        
        left = 2 * i + 1
        right = 2 * i + 2
        mx = i
        
        if left < n and self.temp[mx] < self.temp[left]:
            mx = left
            
        if right < n and self.temp[mx] < self.temp[right]:
            mx = right
            
        if mx == i:
            return 
        
        else:
            temp = self.temp[mx]
            self.temp[mx] = self.temp[i]
            self.temp[i] = temp
        
            self.max_heapify(n,  mx)
                
```

```python
# Using Min_heap

class Solution:
    def heapSort(self, arr):
        #code here
        n = len(arr)
        
        self.build_heap(n, arr) # Step 1: Build Heap
        
        # 2. Extract elements one by one
        for i in range(n - 1, 0, -1):
            # Swap root (min) with the last element (i)
            arr[0], arr[i] = arr[i], arr[0]
            
            # Heapify the reduced heap (size i)
            self.min_heapify(i, 0, arr)

        arr.reverse()
        return arr
        
    def build_heap(self, n, arr): # Build_Heap is done on a unsorted array to bring the max/min element to the root
        
        for i in range((n // 2) - 1, -1, -1):
            
            self.min_heapify(n, i, arr)
        
    def min_heapify(self, n, i, arr): # Heapify is done on a index
        
        left = 2 * i + 1
        right = 2 * i + 2
        mn = i
        
        if left < n and arr[mn] > arr[left]:
            mn = left
            
        if right < n and arr[mn] > arr[right]:
            mn = right
            
        if mn == i:
            return 
        
        else:
            temp = arr[mn]
            arr[mn] = arr[i]
            arr[i] = temp
        
            self.min_heapify(n,  mn, arr)
                
```

```python
import heapq
class Solution:
    def nearlySorted(self, arr, k):  
        #code here
        n = len(arr)
        
        temp = []

        # Add first k-elements
        for i in range(k):
            temp.append(arr[i])
            
        heapq.heapify(temp)
        
        mn = 0
        l = 0
        for i in range(k, n):
            
            mn = heapq.heappushpop(temp, arr[i])
            arr[l] = mn
            l += 1
            # print(mn, arr, temp)
            
        while len(temp) > 0:
            
            ele = heapq.heappop(temp)
            arr[l] = ele
            l += 1            
        # arr = arr[n // 2 ::]
        # print(arr)
        return arr
        
        
```
