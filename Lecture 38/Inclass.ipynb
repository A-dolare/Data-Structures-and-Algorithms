```python
class Solution:
    def minCostClimbingStairs(self, cost: List[int]) -> int:
        n = len(cost)
        
        def solve(n, cost):

            if n == 0:
                return cost[0]
            if n == 1:
                return cost[1]

            c = cost[n] if n != len(cost) else 0
            c += min(solve(n - 1, cost), solve(n - 2, cost))

            return c

        return solve(n, cost)
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        if n == 1:
            return nums[0]
        if n == 2:
            return max(nums[0], nums[1])

        # dp = [-1] * n

        def traverse(house):
            n = len(house)
            if n == 0:
                return 0
            elif n == 1:
                return house[0]
            
            dp = [-1] * (n)
            

            def heist(ind):

                if ind == 0:
                    return house[0]
                if ind == 1:
                    return max(house[0], house[1])

                if dp[ind]  != -1:
                    return dp[ind] 

                dp[ind] = max(heist(ind - 1), house[ind] + heist(ind - 2))

                return dp[ind] 

            return heist(n - 1)

                
        return max(traverse(nums[1:]), traverse(nums[:-1]))
```

```python
# Time Limit Exceeded
class Solution:
    def deleteAndEarn(self, nums: List[int]) -> int:
        maxi = max(nums)
        points = [0] * (maxi + 1)

        for num in nums:
            points[num] += num

        def max_points(arr, n):
            if n == 0:
                return arr[0]
            if n == 1:
                return max(arr[1], arr[0])

            return max(max_points(arr, n - 1), arr[n] + max_points(arr, n - 2))

        return max_points(points, maxi)
        
        
```
