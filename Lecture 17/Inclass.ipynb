```python
class Solution:
    def search(self, arr, key):
        l, h = 0, len(arr) - 1
        
        while l <= h:
            m = (l + h) // 2
            
            if arr[m] == key:
                return m
            
            # Check if left half is sorted
            if arr[l] <= arr[m]:
                # Is key in the sorted left half?
                if arr[l] <= key < arr[m]:
                    h = m - 1
                else:
                    l = m + 1
            # Otherwise, right half must be sorted
            else:
                # Is key in the sorted right half?
                if arr[m] < key <= arr[h]:
                    l = m + 1
                else:
                    h = m - 1
                    
        return -1   
```

```python
class Solution:
    def findMin(self, arr):
        # code here
        n = len(arr)
        l = 0
        h = n - 1
        ans = 0
        
        while l <= h:
            m = (l + h) // 2
            
            if m - 1 >= 0 and arr[m - 1] > arr[m]:
                ans = m
                break
                
            elif arr[m] <= arr[h]:
                h = m - 1
                
            else:
                l = m + 1
                
        return arr[ans]
```
