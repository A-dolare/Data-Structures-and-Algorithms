```python
# MAP FUNTION BASICS
# The map function is a built-in function in Python that allows you to apply a specific function 
# to each item in an iterable (like a list) and returns a map object (which is an iterator). 
# The syntax for the map function is as follows:
# map(function, iterable, ...)

# Here, 'function' is the function you want to apply to each item in the iterable, 
# and 'iterable' is the collection of items you want to process.

# Primary Uses
# Transforming Data: It efficiently applies a transformation function to all 
# elements without writing an explicit for loop.
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, numbers)) 
# Output: [1, 4, 9, 16]

# Using Built-in Functions: It is concise for applying existing 
# functions like str.upper, int, or len to a list.
words = 'hello world'
words = ["hello", "world"]
upper_words = list(map(str.upper, words)) 
print(upper_words)
# Output: ['H', 'E', 'L', 'L', 'O', ' ', 'W', 'O', 'R', 'L', 'D']
# Output: ['HELLO', 'WORLD']

# Processing Multiple Iterables: It can take multiple iterables and apply 
# a function that accepts multiple arguments, processing them in parallel.
list1 = [1, 2, 3]
list2 = [10, 20, 30]
sums = list(map(lambda x, y: x + y, list1, list2)) 
# Output: [11, 22, 33]

# Type Conversion: Commonly used to convert input strings to integers or floats 
# in one line (frequently seen in competitive programming).
# Converts "1 2 3" into [1, 2, 3]

nums = list(map(int, "1 2 3".split())) 
print(nums)  # Output: [1, 2, 3]
X1,X2,X3 = map( str, "1 2 3".split())
print(X1, X2, X3)  # Output: 1 2 3
print(type(X1), type(X2), type(X3))  # Output: <class 'str'> <class 'str'> <class 'str'>

# Taking Input: It is often used to take multiple inputs in a single line and convert them to the desired type.
# For example, to take three integers as input:
Y1, Y2, Y3 = map(int, input("Enter three integers: ").split())
print(Y1, Y2, Y3)  # Output will depend on user input3


```

```python
def buildPrefix(mat):

    n = len(mat)

    pref = [[0] * n for _ in range(n)]
    # Converting the forest: '*' = 1 and '.' = 0 and simultaneously building the prefix sum

    # PSum[i][j] = PSum[i-1][j] + PSum[i][j-1] - PSum[i-1][j-1] + mat[i][j]

    for i in range(n):
        for j in range(n):

            curr = 1 if mat[i][j] == '*' else 0

            top = pref[i-1][j] if i > 0 else 0

            left = pref[i][j-1] if j > 0 else 0

            diag = pref[i-1][j-1] if i > 0 and j > 0 else 0

            pref[i][j] = top + left - diag + curr

    return pref


def query(pref, x1, y1, x2, y2):

    # processing the querries for the number of trees in the given coordinates (x1, y1) and (x2, y2)

    # prefixSum = PSum[x2][y2] - PSum[x1 - 1][y2] - PSum[x2][y1 - 1] + PSum[x1 - 1][y1 - 1]
    
    total = pref[x2][y2]

    top = pref[x1-1][y2] if x1 > 0 else 0

    left = pref[x2][y1-1] if y1 > 0 else 0

    diag = pref[x1-1][y1-1] if x1 > 0 and y1 > 0 else 0

    return total - top - left + diag


def main():

    n, q = map(int, input().split())

    mat = []

    # Taking input for the forest in the form of a matrix
    for _ in range(n):
        mat.append(list(input()))

    pref = buildPrefix(mat)

    # processing querries 
    for _ in range(q):

        x1, y1, x2, y2 = map(int, input().split())

        # convert to 0-index
        x1 -= 1
        y1 -= 1
        x2 -= 1
        y2 -= 1

        print(query(pref, x1, y1, x2, y2))


main()
```
