```python


class Solution:
    def inOrderSuccessor(self, root, x):
        
        def successor(node, target):
            # Base Case: Empty node
            if not node:
                return None
            
            # Case 1: Current node is greater than target
            # This node is a potential successor. 
            # We must check the left subtree to see if there is a closer (smaller) successor.
            if node.data > target.data:
                left_result = successor(node.left, target)
                # If left subtree found a successor, return it. 
                # Otherwise, the current node is the successor.
                return left_result if left_result else node
            
            # Case 2: Current node is smaller or equal
            # The successor must be in the right subtree.
            else:
                return successor(node.right, target)

        # Call the helper and return its result
        return successor(root, x)   
```

```python
# Optimised - using count variable and performing inorder traversal and keeping track 
#             of the ith element when count == k return

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def kthSmallest(self, root: Optional[TreeNode], k: int) -> int:

        self.count = 0
        ele = 0

        def inorder(root, k):

            nonlocal ele

            if not root:
                return

            inorder(root.left, k)

            self.count += 1
            if self.count == k:
                ele = root.val
                return 
            # print(self.count)
            
            inorder(root.right, k)

        inorder(root, k)
        return ele
```
