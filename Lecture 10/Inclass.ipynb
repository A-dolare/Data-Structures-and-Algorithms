```python
a, b, c, trgt = map(int, input().split())

oprtrs = ['*', '+', '-']

result = float('inf')

def solve(a, b, c, trgt, oprtrs, result):
    
    for op1 in oprtrs:
        for op2 in oprtrs:

            if op1 == op2:
                continue
            
            if op1 == '*' and op2 != '*':

                left = a * b

                if op2 == '+':
                    result = left + c
                    if result == trgt:
                        print('YES')
                        return

                elif op2 == '-':
                    result = left - c
                    if result == trgt:
                        print('YES')
                        return

            elif op2 == '*' and op1 != '*':

                right = b * c

                if op1 == '+':
                    result = a + right
                    if result == trgt:
                        print('YES')
                        return

                elif op1 == '-':
                    result = a - right
                    if result == trgt:
                        print('YES')
                        return

            else:
                continue

            if op1 == '+' and op2 == '-':
                result = a + b - c
                if result == trgt:
                        print('YES')
                        return
            
            else:
                result = a - b + c
                if result == trgt:
                        print('YES')
                        return
            # print(result)
    if result != trgt:
        print('NO')
    

solve(a, b, c, trgt, oprtrs, result)
```
