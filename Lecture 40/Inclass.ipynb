```python
class Solution:
    def minPathSum(self, grid):

        n = len(grid)
        m = len(grid[0])
        self.dp = [[-1] * (m + 1) for _ in range(n + 1)]

        return fun(0, 0, grid, n, m)

    def fun(i, j, grid, n, m):

        # base case
        if i == n and j == m:
            return grid[n][m]

        if i < 0 or j < 0 or i >= n or j >= m:
            return float('inf')

        self.dp[i][j] = grid[i][j] + min(func(i, j + 1, grid, n, m), func(i + 1, j, grid, n, m))
    
        return self.dp[i][j]
```
