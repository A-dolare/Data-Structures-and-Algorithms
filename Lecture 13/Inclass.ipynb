```python
# binary number definition:
a = 0b101
print(a)  # Output = 5

# Common Bit Manipulation Techniques

# Beyond basic operators, several standard patterns are used in DSA:

# 1. Checking Parity (Even/Odd)
# Instead of using the modulo operator (%), you can check the least significant bit (LSB).

def is_even(n):
    return (n & 1) == 0

# 2. Multiplying or Dividing by Powers of Two
# Bit shifting is faster than arithmetic multiplication or division for powers of two.

# multiply_by_4 = n << 2  # n * 4
# divide_by_4 = n >> 2    # n // 4

# 3. Setting, Clearing, and Toggling Bits
# You can manipulate specific bits at position k (0-indexed) using masks:

# Set bit: n | (1 << k)
# Clear bit: n & ~(1 << k)
# Toggle bit: n ^ (1 << k)
# Check bit: (n & (1 << k)) != 0 

# 4. Checking Power of Two
# A number is a power of two if it has exactly one bit set. This can be checked efficiently:

def is_power_of_two(n):
    return n > 0 and (n & (n - 1)) == 0

# 5. Swapping Two Numbers
# XOR allows swapping variables without a temporary variable:

# a = a ^ b
# b = a ^ b
# a = a ^ b

# 6. Counting Set Bits (Hamming Weight)
# To count the number of 1s in an integer, repeatedly clear the lowest set bit:

def count_set_bits(n):
    count = 0
    while n:
        n &= (n - 1)
        count += 1
    return count

# Alternatively, Python offers a concise one-liner: bin(n).count('1'). 

# 7. Extracting a Range of Bits
# To isolate a sequence of bits (e.g., bits start to end), create a mask of consecutive 1s, shift it,
#  and apply AND. 

# Logic: (number >> start) & ((1 << (end - start + 1)) - 1)
# Alternative: Create a full mask first: mask = ((1 << width) - 1) << start, then (number & mask) >> start.
def extract_bits(n, start, width):
    return (n >> start) & ((1 << width) - 1)

n = 50       # Binary: 00110010
start = 2
width = 3

# Steps:
# 1. 50 >> 2 = 12 (Binary: 00001100)
# 2. (1 << 3) - 1 = 7 (Binary: 00000111)
# 3. 12 & 7 = 4 (Binary: 00000100)

print(f"Extracted value: {extract_bits(n, start, width)}") 
# Output: 4   

# Example: Extract 3 bits starting from position 2 in 170 (10101010)
# Bits at 2,3,4 are '101' (5)

# 8. Clearing Multiple Specific Bits
# To clear several non-consecutive bits at once, create a mask with 0s at target positions and 1s elsewhere, then use AND. 

# Logic: number & ~mask (where mask has 1s at bits to clear).
# Clear bits 0 and 2 from 15 (1111)
mask = 0b101  # Bits to clear
result = 15 & ~mask  # 1111 & 010 -> 1010 (10)

# Unique Built-in Methods (Python 3.10+):

# Python introduces specific integer methods that act as optimized bit functions, 
# which are not standard keywords in many other languages:

int.bit_count(): #Returns the number of set bits (Hamming weight) in an integer. 
# This replaces the need for manual loops or bin(n).count('1').
# Example: (123).bit_count() returns 5.
int.bit_length(): #Returns the number of bits required to represent an integer in binary, 
                    #excluding the sign and leading zeros.
# Example: (8).bit_length() #returns 4 (since $8$ is 1000).
int.to_bytes() / int.from_bytes(): #Native methods to convert integers directly to byte arrays 
                                    # and vice versa, handling endianness and signedness explicitly. 

# Common Usage Patterns

# Since the 0b prefix is often unnecessary for display or further string processing,
#  it is commonly removed using slicing or alternative formatting methods:

# 1. Removing the Prefix
# Use string slicing [2:] to strip the first two characters.

binary_str = bin(10)[2:]  # Returns '1010'

# 2. Alternative: format() or f-strings
# For cleaner code without slicing, use the format specifier 'b.

binary_str = format(10, 'b')  # Returns '1010'
binary_str = f'{10:b}'        # Returns '1010'

# 3. Zero Padding
# To ensure a fixed width (e.g., 8 bits), combine with zfill() or use format specifiers.

# Using bin()
padded = bin(10)[2:].zfill(8)  # Returns '00001010'

# Using format()
padded = format(10, '08b')     # Returns '00001010'

```
