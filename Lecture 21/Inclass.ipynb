Queue Implementation using simple lists

TC (for enqueue) : O(1)

TC (for dequeue) : O(n)

```python
# Initialize an empty queue
queue = []

# --- Multiple Insertions (Enqueue) ---
print("Enqueuing elements: 'A', 'B', 'C', 'D'")
queue.append('A')
queue.append('B')
queue.append('C')
queue.append('D')
print(f"Queue state: {queue}\n")

# --- Multiple Deletions (Dequeue) ---
# Elements are removed in the same order of insertion
print("Dequeuing elements:")
print(f"Dequeued: {queue.pop(0)}")  # Removes 'A' (shifts B, C, D)  
print(f"Queue state: {queue}\n")
print(f"Dequeued: {queue.pop(0)}")  # Removes 'B' (shifts C, D)
print(f"Queue state: {queue}\n")

# Enqueue more elements
queue.append('E')
queue.append('F')
print(f"Enqueued 'E', 'F'. Queue state: {queue}")
print(f"Final Dequeue: {queue.pop(0)}") # Removes 'C'
print(f"Final Queue: {queue}")   
```

Queue Implementation using collections dequeue()

TC (for enqueue) : O(1)

SC (for dequeue) : O(1)

```python
from collections import deque

# Initialize
q = deque()

# Enqueue
q.append(1)
q.append(2)
q.append(3)

# Size
print(f"Size: {len(q)}")  # Output: 3

# Peek (view front without removing)
if q:
    print(f"Front: {q[0]}")  # Output: 1

q.appendleft(4)

# Dequeue
item = q.popleft()
print(f"Dequeued: {item}")  # Output: 1

# Check Empty
print(f"Is Empty: {not q}") # Output: False 


print(q)
```

Stacks Implentation Using Simple Lists

The most common and intuitive way to implement a stack is using Python's built-in list.  Lists support append() to push elements and pop() to remove them from the end, adhering to the Last-In-First-Out (LIFO) principle. 

Pros & Cons

Pros: Simple, requires no imports, and allows O(1) random access. 

Cons: Lists are dynamic arrays. When the allocated memory block is full, Python must allocate a larger block and copy elements, causing occasional O(n) delays during append() operations.  This makes lists less ideal for very large stacks or high-performance scenarios. 

```python
stack = []

# Push operations
stack.append('A')
stack.append('B')
stack.append('C')

# Peek (view top element)
top_element = stack[-1] if stack else None

# Pop operation
if stack:
    popped = stack.pop()

# Check if empty
is_empty = len(stack) == 0   
```

Stacks Implementation Using Collections.deque()

The deque (double-ended queue) class from the collections module is generally the recommended approach for implementing stacks in Python.  It is implemented as a doubly-linked list, providing consistent O(1) time complexity for both append and pop operations. 

Pros & Cons

Pros: Highly optimized for adding/removing elements from ends; consistent O(1) performance regardless of stack size; thread-safe for simple append/pop operations. 

Cons: Slightly slower than lists for random access (though random access is rarely needed in strict stack usage); requires importing the collections module. 

```python
from collections import deque

stack = deque()

# Push operations
stack.append('A')
stack.append('B')
stack.append('C')

# Peek
top_element = stack[-1] if stack else None

# Pop operation
if stack:
    popped = stack.pop()   
```

```python
# Queue using deqeue
from collections import deque

class myQueue:

    def __init__(self):
        # Initialize your data members
        self.q2 = deque()

    def enqueue(self, x):
        # Implement the enqueue operation
        self.q2.append(x)
        
    def dequeue(self):
        # Implement the dequeue operation
        
        if len(self.q2) == 0:
            return -1
        
        self.q2.popleft()

    def front(self):
        # Return the front element of the queue
        
        if len(self.q2) == 0:
            return -1
        
        item = self.q2[0]
        return item

    def size(self):
        # Return the current size of the queue
        n = len(self.q2)
        return n
```

Input Stack (s1): Acts as the entry point. Every enqueue operation simply pushes an element onto s1. This maintains elements in LIFO (Last-In-First-Out) order relative to insertion.

State: [1, 2, 3] (Top is 3, the newest element).

Output Stack (s2): Acts as the exit point. It holds elements in reverse order compared to s1.

When s2 is empty and a dequeue or front is requested, the algorithm transfers all elements from s1 to s2.

The Flip: Popping from s1 (getting 3, then 2, then 1) and pushing to s2 results in s2 having [3, 2, 1] where 1 is now at the top.

This makes the oldest element (the first one entered) accessible at the top of s2.

| Step | Operation | Action Details | `s1` (Input Stack) | `s2` (Output Stack) | Output / Return |
|:---:|:---|:---|:---:|:---:|:---:|
| 1 | `q.enqueue(1)` | Push `1` to `s1` | `[1]` | `[]` | None |
| 2 | `q.enqueue(2)` | Push `2` to `s1` | `[1, 2]` | `[]` | None |
| 3 | `q.enqueue(3)` | Push `3` to `s1` | `[1, 2, 3]` | `[]` | None |
| 4 | `print(q.s1)` | Print current `s1` | `[1, 2, 3]` | `[]` | `[1, 2, 3]` |
| 5 | `q.front()` | `s2` empty → Transfer all `s1` to `s2` (reversed). Peek top of `s2`. | `[]` | `[3, 2, 1]` | **1** |
| 6 | `q.dequeue()` | `s2` not empty → Pop top of `s2`. | `[]` | `[3, 2]` | **1** |
| 7 | `q.enqueue(4)` | Push `4` to `s1` | `[4]` | `[3, 2]` | None |
| 8 | `print(q.s1)` | Print current `s1` | `[4]` | `[3, 2]` | `[4]` |
| 9 | `print(q.s2)` | Print current `s2` | `[4]` | `[3, 2]` | `[3, 2]` |
| 10 | `q.dequeue()` | `s2` not empty → Pop top of `s2`. | `[4]` | `[3]` | **2** |
| 11 | `q.dequeue()` | `s2` not empty → Pop top of `s2`. | `[4]` | `[]` | **3** |
| 12 | `q.dequeue()` | `s2` empty → Transfer `s1` to `s2`. Pop top of `s2`. | `[]` | `[4]` → `[]` | **4** |
| 13 | `q.dequeue()` | Both empty → Return error code. | `[]` | `[]` | **-1** |

```python
from collections import deque

class myQueue:
    def __init__(self):
        # Initialize two deques to act as stacks
        self.s1 = deque()  # Input Stack
        self.s2 = deque()  # Output Stack

    def enqueue(self, x):
        # Push to Input Stack (O(1))
        self.s1.append(x)
        
    def dequeue(self):
        # If Output Stack is empty, transfer all elements from Input Stack
        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())
        
        # If both are empty, queue is empty
        if not self.s2:
            return -1
            
        # Pop from Output Stack (O(1) amortized)
        return self.s2.pop()

    def front(self):
        # Same logic as dequeue, but peek instead of pop
        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())
        
        if not self.s2:
            return -1
            
        return self.s2[-1]

    def size(self):
        # Total elements in both stacks
        return len(self.s1) + len(self.s2)
    

q = myQueue()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
print(q.s1)
print(f"Front: {q.front()}")      # Output: 1
print(f"Dequeue: {q.dequeue()}")  # Output: 1
q.enqueue(4)
print(q.s1)
print(q.s2)
print(f"Dequeue: {q.dequeue()}")  # Output: 2
print(f"Dequeue: {q.dequeue()}")  # Output: 3
print(f"Dequeue: {q.dequeue()}")  # Output: 4
print(f"Dequeue: {q.dequeue()}")  # Output: -1 (Empty)

```
