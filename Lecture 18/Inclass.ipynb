```python

# Implement the painter_partition method 
def is_feasible(arr, x, p):
    n = len(arr)
    
    time_taken = 0
    painters = 1
    
    for i in range(n):
        
        if arr[i] > x:
            return False
            
        time_taken += arr[i]
        
        if time_taken > x:
            painters += 1
            
            time_taken = arr[i]
            
            if painters > p:
                return  False
                
    return True
    
def painter_partition(boards, k):
    
    l = max(boards)
    h = sum(boards)
    ans = float('inf')
    
    while l <= h:
        m = (l + h) // 2
        if is_feasible(boards, m, k):
            h = m - 1
            ans = m
        else:
            l = m + 1
            
    return ans
    

def main():
    t = int(input())  
    for _ in range(t):
        N, k = map(int, input().split())
        boards = list(map(int, input().split()))
        print(painter_partition(boards, k))

if __name__ == "__main__":
    main() 
```
