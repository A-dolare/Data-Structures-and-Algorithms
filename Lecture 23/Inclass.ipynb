```python
class Solution:
    # Function to calculate factorial of a number.
    def factorial(self, n: int) -> int:
        # code here
        if n < 2:
            return 1
        
        return n * self.factorial(n - 1)
```

```python
class Solution:
    def sumOfDigits(self, n):
        # Base Case: If n is 0, return 0 to stop recursion
        if n == 0:
            return 0
        
        # Recursive Step: Add last digit to the sum of the remaining digits
        return (n % 10) + self.sumOfDigits(n // 10) 
```

```python
class Solution:
    def countDigits(self, n):
        # code here
        
        if n == 0:
            return 0
        
        return 1 + self.countDigits(n // 10)
        # count += 1
```

```python
class Solution:
    def RecursivePower(self, n, p):
        # code here
        
        if n == 1:
            return 1
        
        if p == 0:
            return 1
        
        # n ^ p = (n ^ p//2) * (n ^ p//2)
        temp = self.RecursivePower(n, p // 2)
        
        # print(n, p, temp)
        
        if p % 2 == 0:
            return temp * temp
        else:
            return n * temp * temp
        
```
