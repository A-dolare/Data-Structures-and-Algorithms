```python
import heapq
class Solution:
    def mergeArrays(self, mat):
        # Code here
        n = len(mat)
        if n == 0:
            return []
        
        m = len(mat[0])
        
        ans = []
        
        min_heap = []
        
        col = 0
        
        for i in range(n):
            heapq.heappush(min_heap, (mat[i][0], i, col))
        # print(min_heap)
        
        while min_heap:
            curr, row, col = heapq.heappop(min_heap)
            ans.append(curr)
            
            if col + 1 < m:
                heapq.heappush(min_heap, (mat[row][col + 1], row, col + 1))
                
        return ans
            
```
