```python
from collections import deque
class Solution:
    def rearrangeQueue(self, q):
        #code here

        n = len(q)

        if n == 0:
            return q

        q1 = deque()
        q2 = deque()

        for i in range(n):
            if i < n // 2:
                q1.append(q[i])
            else:
                q2.append(q[i])

        q.clear()

        for i in range(n // 2):
            q.append(q1[i])
            q.append(q2[i])

        return q
```
