```python

from typing import List


class Solution:
    def findPair(self, arr: List[int], x: int) -> int:
        # code here
        
        arr.sort()
        
        i=0
        j=1
        
        while  j<len(arr) and  i<len(arr):
            
            if i==j:
                j+=1
                continue
            
            diff=abs(arr[i]-arr[j])
            
            if diff<x:
                j+=1
            elif diff>x:
                i+=1
            else:
                return True
                
        return False
    
soln=Solution()
print(soln.findPair([1,2,3,4,2,1,43,2,1], 23))
        
```

```python
class Solution:

    def findPair(self, arr, x):

        seen = set()

        for num in arr:

            # check if required pair exists
            if (num + x) in seen:
                return True

            if (num - x) in seen:
                return True

            # store current number
            seen.add(num)

        return False
    
soln=Solution()
print(soln.findPair([1,2,3,4,2,1,43,2,1], 5))
```

```python

class Solution:

    def minDifference(self, arr):

        n = len(arr)

        # calculate total sum
        total_sum = sum(arr)

        # initially:
        # left side has first element
        left_sum = arr[0]

        # right side has remaining elements
        right_sum = total_sum - left_sum

        # initial answer
        ans = abs(left_sum - right_sum)

        # move partition
        for i in range(1, n):

            # include current element in left side
            left_sum += arr[i]

            # remove current element from right side
            right_sum -= arr[i]

            # calculate difference
            diff = abs(left_sum - right_sum)

            # update minimum answer
            ans = min(ans, diff)

        return ans
soln=Solution()
print(soln.minDifference([1,2,3,4,5,6]))

```
