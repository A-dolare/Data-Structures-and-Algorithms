```python
class Solution:
    def findUnique(self, arr):
        # code here
        xor = 0
        for num in arr:
            xor ^= num
        return xor
        
```

```python
class Solution:
    def missingNumber(self, nums: List[int]) -> int:
        
        n = len(nums)
        total = ((n + 1) * n) // 2

        for n in nums:
            total -= n

        return total
```

```python
class Solution:
    def missingNumber(self, nums: List[int]) -> int:
        n = len(nums) 
        xor = 0

        for i in range(n + 1):
            xor ^= i
        for i in range(n):
            xor ^= nums[i] 

        return xor

        
```

```python
# Brain's Algo
class Solution:
	def setBits(self, n):
		# code here
		count = 0 
		num = n
		while n > 0:
		    n = num & (num - 1)
		    count += 1
		    num = n
		
		return count
```
