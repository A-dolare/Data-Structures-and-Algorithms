```python
class Solution:
    def isPrime(self, n):
        if n==1:
            return False
        if n==2 or n==3:
            return True
        if n%2==0 or n%3==0:
            return False
        
        i=5
        while i*i<=n:
            if n%i==0 or n%(i+2)==0:
                return False
            i+=6
        return True

soln = Solution()
print('is prime number' if soln.isPrime(17) ==True else "Not prime")
```

```python
class Solution:
    
    def reverse(self , arr , l, u):
        
        while l<u:
            
            temp = arr[l]
            arr[l] = arr[u]
            arr[u] = temp
            
            l+=1
            u-=1
        return arr
            
        
    def rotateArr(self, arr, d):
        
        n = len(arr)
        
        d = d % n
        
        
        self.reverse(arr, 0 ,d - 1)
        self.reverse(arr, d ,n - 1)
        self.reverse(arr, 0, n - 1)
        
        
        
        return arr
```
