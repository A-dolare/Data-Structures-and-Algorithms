```python
# Kth Largest using max_heap

#User function Template for python3
import heapq
class Solution:
    #Function to return kth largest element from an array.
    def kthLargest(self,arr,k):
        n = len(arr)
        
        max_heap = [-n for n in arr]
        
        heapq.heapify(max_heap)        
        
        for i in range(k):
            mx = heapq.heappop(max_heap)
            # print(-mx)
            
        return -mx
        
```

```python
import heapq
class Solution:
    def findRelativeRanks(self, score: List[int]) -> List[str]:

        n = len(score)

        max_heap = [-n for n in score]  
        heapq.heapify(max_heap) 

        mx = heapq.heappop(max_heap)
        # print(mx)
        idx = score.index(-mx)
        score[idx] = "Gold Medal"

        if len(max_heap) == 0:
            return score

        mx = heapq.heappop(max_heap)
        idx = score.index(-mx)
        score[idx] = "Silver Medal"

        if len(max_heap) == 0:
            return score

        mx = heapq.heappop(max_heap)
        idx = score.index(-mx)
        score[idx] = "Bronze Medal"

        if len(max_heap) == 0:
            return score

        i = 4

        while len(max_heap) > 0:
            mx = heapq.heappop(max_heap)
            idx = score.index(-mx)

            score[idx] = str(i)
            i += 1
            

        return score
```

```python
import heapq
import math
class Solution:
    def pickGifts(self, gifts: List[int], k: int) -> int:

        max_heap = [-n for n in gifts]

        heapq.heapify(max_heap)

        mx = 0
        for i in range(k):
            mx = heapq.heappop(max_heap) * -1
            heapq.heappush(max_heap, -math.floor(math.sqrt(mx)))
        # print(max_heap)
        return -sum(max_heap)
```
