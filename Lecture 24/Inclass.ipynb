```python
class Solution:
    def targetSumComb(self, arr, target):
        # arr.sort()
        ans = []
        temp = []
        self.fun(arr, 0, temp, ans, target)
        return ans

    def fun(self, arr, i, temp, ans, target):
        # Base case: target reached
        if target == 0:
            ans.append(temp[:])
            return
        
        # Base case: invalid path
        if i >= len(arr) or target < 0:
            return
        
        # Include arr[i] (stay at same index to allow reuse)
        if arr[i] <= target:
            temp.append(arr[i])
            self.fun(arr, i, temp, ans, target - arr[i])  # Note: i, not i+1 because an element can occur multiple times
            temp.pop()
        
        # Exclude arr[i] and move to next index
        self.fun(arr, i + 1, temp, ans, target)   
```
