```python
# Adjacency matrix for undirected graph
# link : https://ideone.com/fsjYS8

V, E = map(int, input().split())

adj_mat = [[0] * V for _ in range(V)]

for _ in range(E):
    
    node1, node2 = map(int, input().split())
    
    adj_mat[node1][node2] = 1
    adj_mat[node2][node1] = 1
    
    
print(adj_mat)
```

```python
# Adjacency matrix for weighted undirected graph

V, E = map(int, input().split())

mat = [[0] * V for _ in range(V)]

for _ in range(E):
    src, dst, wght = map(int, input().split())

    mat[src][dst] = wght
    mat[dst][src] = wght

print(mat)
```

1. Adjacency List for Directed and unweighted graph

```python
def add_edge(adj_list, u, v):
    adj_list[u].append(v)

def print_list(adj_list):

    for i in range(len(adj_list)):
        print(f'{i}:', end = '')
        for j in range(len(adj_list[i])):
            print(adj_list[i][j], sep=',')
        print()
        

def main():
    V = 3
    adj_list = [[] for i in range(V)]

    add_edge(adj_list, 0, 1)
    add_edge(adj_list, 1, 2)
    add_edge(adj_list, 2, 0)

    print_list(adj_list)

main()   
```
