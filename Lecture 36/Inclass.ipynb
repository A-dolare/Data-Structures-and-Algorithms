```python
from collections import deque
class Solution:
    def is_valid(self, rr, cc, n, m):
        return True if 0 <= rr < n and 0 <= cc < m else False
        
    def orangesRot(self, mat):
        # code here
        n = len(mat)
        m = len(mat[0])
        q = deque()
        
        dr = [-1, 1, 0, 0]
        dc = [0, 0, -1, 1]
        
        fresh = 0
        
        for i in range(n):
            for j in range(m):
                if mat[i][j] == 1:
                    fresh += 1
                    
                if mat[i][j] == 2:
                    q.append((i, j))
                    
        if fresh == 0:
            return 0
                    
        time = 0
        
        while len(q) > 0:
            rotten_count = len(q)
            
            while rotten_count > 0:
                
                curr = q.popleft()
                
                for i in range(4):
                    rr = curr[0] + dr[i]
                    cc = curr[1] + dc[i] 
                    
                    if self.is_valid(rr, cc, n, m) and mat[rr][cc] not in [0, 2]:
                        
                        mat[rr][cc] = 2
                        q.append((rr, cc))
                        fresh -= 1
                        
                rotten_count -= 1
                
            time += 1
            
            if fresh == 0:
                return time
                
                
        if fresh == 0:
            return time
        else:
            return -1
                        
                        
                        
                
```

```python
from collections import deque
class Solution:
    def topoSort(self, V, edges):
        # Code here
        q = deque()
        ans = []
        
        indegree = [0] * V

        # converting edges representation -> Adjacency List
        adj = [[] for _ in range(V)]
        for edge in edges:
            adj[edge[0]].append(edge[1])
            indegree[edge[1]] += 1

        # add vertices with indegree == 0 to queue and ans
        for i in range(V):
            if indegree[i] == 0:
                q.append(i)
                ans.append(i)
                
        while len(q) > 0:
            curr = q.popleft()
            
            # reduce the indegrees of all the adjacent vertices
            
            ngbrs = adj[curr]
            for vert in ngbrs:
                indegree[vert] -= 1
                if indegree[vert] == 0:
                    q.append(vert)
                    ans.append(vert)
                    
        return ans
```
