```python
class Solution:
    def MaxSubarraySum(self , arr):
        
        presum = arr[0]
        ans = presum 
        
        for i in range(1, len(arr)):
            
            presum = max( presum + arr[i] , arr[i] )
            ans = max( presum , ans )
        
        return ans
        
    def MinSubarraySum(self , arr):
        
        presum = arr[0]
        ans = presum 
        
        for i in range(1, len(arr)):
            
            presum = min( presum + arr[i] , arr[i] )
            ans = min( presum , ans )
        
        return ans
            
        
    def maxCircularSum(self, arr):
        # code here
        
        ans1 = self.MaxSubarraySum ( arr )
        
        totalSum = sum( arr )
        
        ans2 = totalSum - self.MinSubarraySum( arr )
        
        if totalSum == self.MinSubarraySum( arr ):  #Since the problem requires a non-empty subarray
            return ans1
        
        return max( ans1 , ans2 )
        
```

Dry Run: Maximum Sum Subarray of Size at Least K
Input: arr = [-4, -2, 1, -3], k = 2
Expected Output: -1 (from subarray [-2, 1]) 

Phase 1: Kadane's Preprocessing (maxSum Array)
We compute maxSum[i], the maximum subarray sum ending at index i.

| Index (`i`) | `arr[i]` | Calculation (`max(arr[i], currMax + arr[i])`) | `currMax` | `maxSum[i]` |
|:---:|:---:|:---|:---:|:---:|
| 0 | -4 | Initialize | -4 | **-4** |
| 1 | -2 | `max(-2, -4 + -2) = -2` | -2 | **-2** |
| 2 | 1 | `max(1, -2 + 1) = 1` | 1 | **1** |
| 3 | -3 | `max(-3, 1 + -3) = -2` | -2 | **-2** |

Phase 2: Sliding Window & Result Calculation
We slide a window of size k=2 and check two options at each step:

Exact K: Just the current window sum.
At Least K: Current window sum + maxSum before the window started. 
Initialization:

Initial Window (arr[0..1]): [-4, -2]
window_sum = $-4 + (-2) = -6$
result = -6
Iteration Trace:

| Step | `i` | `arr[i]` | Window Elements | `window_sum` Calculation | Option 1: `window_sum` | Option 2: `window_sum + maxSum[i-k]` | New `result` |
|:---:|:---:|:---:|:---:|:---|:---:|:---|:---:|
| **Init** | - | - | `[-4, -2]` | $-4 + (-2)$ | - | - | **-6** |
| **1** | 2 | 1 | `[-2, 1]` | $-6 + 1 - (-4) = \mathbf{-1}$ | `-1` | $-1 + maxSum[0] = -1 + (-4) = -5$ | **max(-6, -1, -5) = -1** |
| **2** | 3 | -3 | `[1, -3]` | $-1 + (-3) - (-2) = \mathbf{-2}$ | `-2` | $-2 + maxSum[1] = -2 + (-2) = -4$ | **max(-1, -2, -4) = -1** |

Final Conclusion
Maximum Sum Found: -1
Subarray: [-2, 1] (Indices 1 to 2)
The algorithm correctly identified that extending the window [1, -3] backwards with maxSum[1] yielded a worse result, so it stuck with the best found so far (-1)

```python
class Solution:
    def maxSumWithK(self, arr, k):
        # code here
        n = len(arr)
        
        # prefix = [0] * n
        # prefix[0] = arr[0]
        # for i in range(1, n):
        #     prefix[i] = prefix[i - 1] + arr[i]
        # print(prefix)
        
        curr_sum = arr[0]
        max_sum = [0] * n
        max_sum[0] = arr[0]
        for i in range(1, n):
            curr_sum = max(arr[i], curr_sum + arr[i])
            max_sum[i] = curr_sum
        # print(max_sum)
            
        result = sum(arr[:k])
        window_sum = result
        
        for i in range(k, n):
            window_sum = window_sum + arr[i] - arr[i - k]
            
            result = max(result, window_sum)
            
            result = max(result, window_sum + max_sum[i - k])
            # print(result, window_sum)
            
        return result
```
